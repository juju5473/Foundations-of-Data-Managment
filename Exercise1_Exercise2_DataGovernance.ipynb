### 2.a. Data Quality Assessment

#### Completeness Analysis

````python
# Data Quality Log
quality_log = []

def log_quality_issue(table, issue_type, description, count, severity):
    """Log data quality issues"""
    quality_log.append({
        'Timestamp': datetime.now(),
        'Table': table,
        'Issue Type': issue_type,
        'Description': description,
        'Count': count,
        'Severity': severity
    })

# Check completeness for Orders table
print("ORDERS TABLE - COMPLETENESS ANALYSIS")
print("="*80)
missing_data = df_orders.isnull().sum()
missing_pct = (df_orders.isnull().sum() / len(df_orders)) * 100

completeness_df = pd.DataFrame({
    'Column': missing_data.index,
    'Missing Count': missing_data.values,
    'Missing %': missing_pct.values
})
completeness_df = completeness_df[completeness_df['Missing Count'] > 0].sort_values('Missing Count', ascending=False)

print(completeness_df)
print(f"\nTotal Records: {len(df_orders)}")
print(f"Columns with Missing Data: {len(completeness_df)}")

# Log completeness issues
for _, row in completeness_df.iterrows():
    severity = 'HIGH' if row['Missing %'] > 5 else 'MEDIUM' if row['Missing %'] > 1 else 'LOW'
    log_quality_issue('Orders', 'Completeness', 
                     f"Missing values in {row['Column']}", 
                     row['Missing Count'], severity)
````

````python
# Check completeness for People table
print("\nPEOPLE TABLE - COMPLETENESS ANALYSIS")
print("="*80)
missing_people = df_people.isnull().sum()
print(missing_people[missing_people > 0])

if missing_people.sum() > 0:
    log_quality_issue('People', 'Completeness', 
                     f"Missing regions in People table", 
                     missing_people.sum(), 'HIGH')
````

````python
# Check completeness for Returns table
print("\nRETURNS TABLE - COMPLETENESS ANALYSIS")
print("="*80)
missing_returns = df_returns.isnull().sum()
print(missing_returns[missing_returns > 0])

if missing_returns.sum() > 0:
    log_quality_issue('Returns', 'Completeness', 
                     f"Missing Order IDs in Returns table", 
                     missing_returns.sum(), 'HIGH')
````

#### Inconsistencies Analysis

````python
print("\nINCONSISTENCY ANALYSIS")
print("="*80)

# Check for date inconsistencies (Ship Date before Order Date)
date_issues = df_orders[df_orders['Ship Date'] < df_orders['Order Date']]
print(f"\n1. Date Inconsistencies (Ship Date < Order Date): {len(date_issues)}")
if len(date_issues) > 0:
    print(date_issues[['Order ID', 'Order Date', 'Ship Date']].head())
    log_quality_issue('Orders', 'Inconsistency', 
                     'Ship Date before Order Date', 
                     len(date_issues), 'HIGH')

# Check for negative quantities
neg_quantity = df_orders[df_orders['Quantity'] <= 0]
print(f"\n2. Negative or Zero Quantities: {len(neg_quantity)}")
if len(neg_quantity) > 0:
    print(neg_quantity[['Order ID', 'Product Name', 'Quantity']].head())
    log_quality_issue('Orders', 'Inconsistency', 
                     'Negative or zero quantities', 
                     len(neg_quantity), 'HIGH')

# Check for discount > 100% or < 0%
invalid_discount = df_orders[(df_orders['Discount'] < 0) | (df_orders['Discount'] > 1)]
print(f"\n3. Invalid Discounts (< 0% or > 100%): {len(invalid_discount)}")
if len(invalid_discount) > 0:
    print(invalid_discount[['Order ID', 'Product Name', 'Discount']].head())
    log_quality_issue('Orders', 'Inconsistency', 
                     'Invalid discount percentages', 
                     len(invalid_discount), 'HIGH')

# Check for negative sales or profit where it seems incorrect
print(f"\n4. Negative Sales: {len(df_orders[df_orders['Sales'] < 0])}")
print(f"5. Negative Profit: {len(df_orders[df_orders['Profit'] < 0])}")
print("   Note: Negative profit can be legitimate due to discounts/costs")
````

#### Redundancies and Duplicates Analysis

````python
print("\nREDUNDANCY AND DUPLICATE ANALYSIS")
print("="*80)

# Check for duplicate Row IDs
duplicate_row_ids = df_orders[df_orders['Row ID'].duplicated(keep=False)]
print(f"\n1. Duplicate Row IDs: {len(duplicate_row_ids)}")
if len(duplicate_row_ids) > 0:
    print(duplicate_row_ids[['Row ID', 'Order ID']].head(10))
    log_quality_issue('Orders', 'Duplication', 
                     'Duplicate Row IDs found', 
                     len(duplicate_row_ids), 'MEDIUM')

# Check for complete duplicate rows
duplicate_rows = df_orders[df_orders.duplicated(keep=False)]
print(f"\n2. Complete Duplicate Rows: {len(duplicate_rows)}")
if len(duplicate_rows) > 0:
    log_quality_issue('Orders', 'Duplication', 
                     'Complete duplicate rows', 
                     len(duplicate_rows), 'HIGH')

# Check for duplicate customers (same name, different ID)
customer_name_counts = df_orders.groupby('Customer Name')['Customer ID'].nunique()
multi_id_customers = customer_name_counts[customer_name_counts > 1]
print(f"\n3. Customers with Multiple IDs: {len(multi_id_customers)}")
if len(multi_id_customers) > 0:
    print("Sample cases:")
    for name in list(multi_id_customers.index)[:5]:
        ids = df_orders[df_orders['Customer Name'] == name]['Customer ID'].unique()
        print(f"  {name}: {ids}")
    log_quality_issue('Orders', 'Duplication', 
                     'Customers with multiple IDs', 
                     len(multi_id_customers), 'MEDIUM')

# Check People table for duplicates
duplicate_people = df_people[df_people.duplicated(subset=['Regional Manager'], keep=False)]
print(f"\n4. Duplicate Regional Managers in People table: {len(duplicate_people)}")
if len(duplicate_people) > 0:
    print(duplicate_people)
    log_quality_issue('People', 'Duplication', 
                     'Duplicate regional managers', 
                     len(duplicate_people), 'HIGH')
````

### 2.b. Additional Calculations and KPIs

````python
print("\nADDITIONAL CALCULATIONS AND KPIs")
print("="*80)

# Create a copy of the dataframe for calculations
df_enhanced = df_orders.copy()

# 1. Cost of Goods Sold (COGS)
# Formula: COGS = Sales - Profit
df_enhanced['COGS'] = df_enhanced['Sales'] - df_enhanced['Profit']
print("\n1. Added COGS (Cost of Goods Sold)")
print("   Formula: Sales - Profit")

# 2. Profit Margin %
# Formula: (Profit / Sales) * 100
df_enhanced['Profit Margin %'] = (df_enhanced['Profit'] / df_enhanced['Sales']) * 100
df_enhanced['Profit Margin %'] = df_enhanced['Profit Margin %'].replace([np.inf, -np.inf], np.nan)
print("\n2. Added Profit Margin %")
print("   Formula: (Profit / Sales) * 100")

# 3. Revenue per Unit
# Formula: Sales / Quantity
df_enhanced['Revenue per Unit'] = df_enhanced['Sales'] / df_enhanced['Quantity']
print("\n3. Added Revenue per Unit")
print("   Formula: Sales / Quantity")

# 4. Discount Amount in dollars
# Formula: Sales / (1 - Discount) * Discount
df_enhanced['Discount Amount'] = (df_enhanced['Sales'] / (1 - df_enhanced['Discount'].replace(0, 0.0001))) * df_enhanced['Discount']
df_enhanced.loc[df_enhanced['Discount'] == 0, 'Discount Amount'] = 0
print("\n4. Added Discount Amount")
print("   Formula: (Sales / (1 - Discount%)) * Discount%")

# 5. List Price (before discount)
# Formula: Sales / (1 - Discount)
df_enhanced['List Price'] = df_enhanced['Sales'] / (1 - df_enhanced['Discount'].replace(0, 0.0001))
df_enhanced.loc[df_enhanced['Discount'] == 0, 'List Price'] = df_enhanced.loc[df_enhanced['Discount'] == 0, 'Sales']
print("\n5. Added List Price")
print("   Formula: Sales / (1 - Discount%)")

# 6. Shipping Time (in days)
df_enhanced['Shipping Days'] = (df_enhanced['Ship Date'] - df_enhanced['Order Date']).dt.days
print("\n6. Added Shipping Days")
print("   Formula: Ship Date - Order Date")

# 7. Time-based attributes
df_enhanced['Order Year'] = df_enhanced['Order Date'].dt.year
df_enhanced['Order Quarter'] = df_enhanced['Order Date'].dt.quarter
df_enhanced['Order Month'] = df_enhanced['Order Date'].dt.month
df_enhanced['Order Month Name'] = df_enhanced['Order Date'].dt.month_name()
df_enhanced['Order Week'] = df_enhanced['Order Date'].dt.isocalendar().week
df_enhanced['Order Day of Week'] = df_enhanced['Order Date'].dt.day_name()
print("\n7. Added Time-based Attributes")
print("   - Order Year, Quarter, Month, Week, Day of Week")

# 8. Year-Quarter and Year-Month for easier grouping
df_enhanced['Year-Quarter'] = df_enhanced['Order Year'].astype(str) + '-Q' + df_enhanced['Order Quarter'].astype(str)
df_enhanced['Year-Month'] = df_enhanced['Order Date'].dt.to_period('M').astype(str)
print("\n8. Added Year-Quarter and Year-Month")

print("\n" + "="*80)
print("Summary of New Columns:")
new_columns = ['COGS', 'Profit Margin %', 'Revenue per Unit', 'Discount Amount', 
               'List Price', 'Shipping Days', 'Order Year', 'Order Quarter', 
               'Order Month', 'Order Month Name', 'Order Week', 'Order Day of Week',
               'Year-Quarter', 'Year-Month']
for col in new_columns:
    print(f"  ✓ {col}")
print(f"\nTotal columns before: {len(df_orders.columns)}")
print(f"Total columns after: {len(df_enhanced.columns)}")
print(f"New columns added: {len(df_enhanced.columns) - len(df_orders.columns)}")
````

### 2.c. Data Cleaning Process and Log

````python
# Create a detailed cleaning log
cleaning_log = []

def log_cleaning_action(action_type, description, records_affected, before_value=None, after_value=None):
    """Log all data cleaning actions"""
    cleaning_log.append({
        'Timestamp': datetime.now(),
        'Action Type': action_type,
        'Description': description,
        'Records Affected': records_affected,
        'Before Value': before_value,
        'After Value': after_value
    })

print("\nDATA CLEANING PROCESS")
print("="*80)
print(f"Starting record count: {len(df_enhanced)}")

# Create a copy for cleaning
df_cleaned = df_enhanced.copy()
initial_count = len(df_cleaned)
````

````python
# STEP 1: Handle Missing Row IDs
print("\nSTEP 1: Handle Missing Row IDs")
print("-" * 80)
missing_row_ids = df_cleaned['Row ID'].isnull().sum()
if missing_row_ids > 0:
    # Generate new Row IDs for missing values
    max_row_id = df_cleaned['Row ID'].max()
    null_mask = df_cleaned['Row ID'].isnull()
    new_ids = range(int(max_row_id) + 1, int(max_row_id) + 1 + null_mask.sum())
    df_cleaned.loc[null_mask, 'Row ID'] = list(new_ids)
    log_cleaning_action('UPDATE', 'Generated new Row IDs for missing values', 
                       missing_row_ids, 'NULL', 'Generated Sequential IDs')
    print(f"✓ Generated {missing_row_ids} new Row IDs")
else:
    print("✓ No missing Row IDs found")
````

````python
# STEP 2: Handle Missing Order IDs
print("\nSTEP 2: Handle Missing Order IDs")
print("-" * 80)
missing_order_ids = df_cleaned['Order ID'].isnull().sum()
if missing_order_ids > 0:
    print(f"⚠ Warning: Found {missing_order_ids} records with missing Order IDs")
    print("Action: These records will be flagged for review but kept in dataset")
    # Create a flag column
    df_cleaned['Data_Quality_Flag'] = ''
    df_cleaned.loc[df_cleaned['Order ID'].isnull(), 'Data_Quality_Flag'] = 'MISSING_ORDER_ID'
    log_cleaning_action('FLAG', 'Flagged records with missing Order IDs', 
                       missing_order_ids, 'NULL', 'FLAGGED_FOR_REVIEW')
else:
    print("✓ No missing Order IDs found")
````

````python
# STEP 3: Handle Missing Postal Codes
print("\nSTEP 3: Handle Missing Postal Codes")
print("-" * 80)
missing_postal = df_cleaned['Postal Code'].isnull().sum()
if missing_postal > 0:
    print(f"Found {missing_postal} missing postal codes")
    # Try to fill based on City and State
    # Create a mapping from known City-State combinations
    postal_mapping = df_cleaned.dropna(subset=['Postal Code']).groupby(['City', 'State'])['Postal Code'].first().to_dict()
    
    filled_count = 0
    for idx, row in df_cleaned[df_cleaned['Postal Code'].isnull()].iterrows():
        key = (row['City'], row['State'])
        if key in postal_mapping:
            df_cleaned.loc[idx, 'Postal Code'] = postal_mapping[key]
            filled_count += 1
    
    if filled_count > 0:
        log_cleaning_action('UPDATE', 'Filled missing Postal Codes based on City-State mapping', 
                           filled_count, 'NULL', 'Filled from mapping')
        print(f"✓ Filled {filled_count} postal codes using City-State mapping")
    
    still_missing = df_cleaned['Postal Code'].isnull().sum()
    if still_missing > 0:
        print(f"⚠ {still_missing} postal codes still missing - keeping as NULL")
        log_cleaning_action('FLAG', 'Postal codes that could not be filled', 
                           still_missing, 'NULL', 'Kept as NULL')
else:
    print("✓ No missing Postal Codes")
````

````python
# STEP 4: Handle Other Missing Values
print("\nSTEP 4: Handle Other Missing Values")
print("-" * 80)

# Ship Date - forward fill from Order Date + average shipping time
missing_ship_date = df_cleaned['Ship Date'].isnull().sum()
if missing_ship_date > 0:
    avg_shipping = df_cleaned['Shipping Days'].mean()
    for idx in df_cleaned[df_cleaned['Ship Date'].isnull()].index:
        df_cleaned.loc[idx, 'Ship Date'] = df_cleaned.loc[idx, 'Order Date'] + pd.Timedelta(days=int(avg_shipping))
    log_cleaning_action('UPDATE', 'Filled missing Ship Dates using average shipping time', 
                       missing_ship_date, 'NULL', f'Order Date + {int(avg_shipping)} days')
    print(f"✓ Filled {missing_ship_date} Ship Dates using average shipping time")

# Ship Mode, Customer Name, Segment, Country/Region, State - fill with 'UNKNOWN'
text_columns = ['Ship Mode', 'Customer Name', 'Segment', 'Country/Region', 'State']
for col in text_columns:
    missing_count = df_cleaned[col].isnull().sum()
    if missing_count > 0:
        df_cleaned[col] = df_cleaned[col].fillna('UNKNOWN')
        log_cleaning_action('UPDATE', f'Filled missing {col} with UNKNOWN', 
                           missing_count, 'NULL', 'UNKNOWN')
        print(f"✓ Filled {missing_count} missing {col} with 'UNKNOWN'")
````

````python
# STEP 5: Handle Duplicate Row IDs
print("\nSTEP 5: Handle Duplicate Row IDs")
print("-" * 80)
duplicate_row_ids = df_cleaned[df_cleaned['Row ID'].duplicated(keep=False)]
if len(duplicate_row_ids) > 0:
    print(f"Found {len(duplicate_row_ids)} records with duplicate Row IDs")
    # Keep first occurrence, reassign Row IDs to duplicates
    max_row_id = df_cleaned['Row ID'].max()
    dup_mask = df_cleaned['Row ID'].duplicated(keep='first')
    new_ids = range(int(max_row_id) + 1, int(max_row_id) + 1 + dup_mask.sum())
    df_cleaned.loc[dup_mask, 'Row ID'] = list(new_ids)
    log_cleaning_action('UPDATE', 'Reassigned Row IDs for duplicates', 
                       dup_mask.sum(), 'Duplicate ID', 'New Unique ID')
    print(f"✓ Reassigned {dup_mask.sum()} duplicate Row IDs")
else:
    print("✓ No duplicate Row IDs found")
````

````python
# STEP 6: Remove Complete Duplicate Rows
print("\nSTEP 6: Remove Complete Duplicate Rows")
print("-" * 80)
before_dedup = len(df_cleaned)
# Consider duplicates based on business keys (exclude Row ID)
business_keys = ['Order ID', 'Product ID', 'Customer ID', 'Order Date']
df_cleaned = df_cleaned.drop_duplicates(subset=business_keys, keep='first')
duplicates_removed = before_dedup - len(df_cleaned)
if duplicates_removed > 0:
    log_cleaning_action('DELETE', 'Removed complete duplicate rows', 
                       duplicates_removed, 'Duplicate', 'Removed')
    print(f"✓ Removed {duplicates_removed} duplicate rows")
else:
    print("✓ No complete duplicate rows found")
````

````python
# STEP 7: Handle Data Inconsistencies
print("\nSTEP 7: Handle Data Inconsistencies")
print("-" * 80)

# Fix Ship Date < Order Date
date_issues = df_cleaned[df_cleaned['Ship Date'] < df_cleaned['Order Date']]
if len(date_issues) > 0:
    # Swap the dates
    for idx in date_issues.index:
        df_cleaned.loc[idx, ['Order Date', 'Ship Date']] = df_cleaned.loc[idx, ['Ship Date', 'Order Date']].values
    log_cleaning_action('UPDATE', 'Corrected Ship Date < Order Date by swapping', 
                       len(date_issues), 'Incorrect order', 'Dates swapped')
    print(f"✓ Corrected {len(date_issues)} date inconsistencies by swapping")
else:
    print("✓ No date inconsistencies found")

# Recalculate Shipping Days after date corrections
df_cleaned['Shipping Days'] = (df_cleaned['Ship Date'] - df_cleaned['Order Date']).dt.days
````

````python
# STEP 8: Clean People Table
print("\nSTEP 8: Clean People Table")
print("-" * 80)

df_people_cleaned = df_people.copy()

# Handle duplicate regional managers
duplicates = df_people_cleaned[df_people_cleaned['Regional Manager'].duplicated(keep=False)]
if len(duplicates) > 0:
    print(f"Found {len(duplicates)} duplicate entries:")
    print(duplicates)
    # Keep the one with Region, remove the one without
    df_people_cleaned = df_people_cleaned.dropna(subset=['Region'])
    # Then remove any remaining duplicates
    df_people_cleaned = df_people_cleaned.drop_duplicates(subset=['Regional Manager'], keep='first')
    log_cleaning_action('DELETE', 'Removed duplicate Regional Managers from People table', 
                       len(duplicates), 'Duplicate', 'Removed')
    print(f"✓ Cleaned {len(duplicates)} duplicate entries")
else:
    print("✓ No duplicates in People table")

print(f"\nPeople table - Final shape: {df_people_cleaned.shape}")
````

````python
# STEP 9: Clean Returns Table
print("\nSTEP 9: Clean Returns Table")
print("-" * 80)

df_returns_cleaned = df_returns.copy()

# Handle missing Order IDs
missing_order_ids_returns = df_returns_cleaned['Order ID'].isnull().sum()
if missing_order_ids_returns > 0:
    print(f"⚠ Found {missing_order_ids_returns} returns with missing Order IDs")
    print("Action: Removing these records as they cannot be linked to orders")
    df_returns_cleaned = df_returns_cleaned.dropna(subset=['Order ID'])
    log_cleaning_action('DELETE', 'Removed returns with missing Order IDs', 
                       missing_order_ids_returns, 'NULL Order ID', 'Removed')
    print(f"✓ Removed {missing_order_ids_returns} records")
else:
    print("✓ No missing Order IDs in Returns table")

# Remove duplicates
before_dedup_returns = len(df_returns_cleaned)
df_returns_cleaned = df_returns_cleaned.drop_duplicates(subset=['Order ID'], keep='first')
dedup_count = before_dedup_returns - len(df_returns_cleaned)
if dedup_count > 0:
    log_cleaning_action('DELETE', 'Removed duplicate returns', 
                       dedup_count, 'Duplicate', 'Removed')
    print(f"✓ Removed {dedup_count} duplicate returns")

print(f"\nReturns table - Final shape: {df_returns_cleaned.shape}")
````

````python
print("\nCREATING NORMALIZED TABLES FOR EXERCISE 2")
print("="*80)

# We'll create normalized tables from the cleaned dataset
````

````python
# 1. CUSTOMER Table (Master Data)
print("\n1. Creating CUSTOMER table (Master Data)...")

customer_table = df_cleaned.groupby('Customer ID').agg({
    'Customer Name': 'first',
    'Segment': 'first',
    'Order Date': ['min', 'max'],
    'Sales': 'sum'
}).reset_index()

customer_table.columns = ['Customer_ID', 'Customer_Name', 'Segment', 
                          'First_Order_Date', 'Last_Order_Date', 'Total_Lifetime_Value']

print(f"  ✓ Created Customer table with {len(customer_table)} unique customers")
print(f"  Columns: {customer_table.columns.tolist()}")
````

````python
# 2. PRODUCT Table (Master Data)
print("\n2. Creating PRODUCT table (Master Data)...")

product_table = df_cleaned.groupby('Product ID').agg({
    'Product Name': 'first',
    'Category': 'first',
    'Sub-Category': 'first',
    'Revenue per Unit': 'mean',  # Average unit price
    'COGS': 'mean'  # Average cost
}).reset_index()

product_table.columns = ['Product_ID', 'Product_Name', 'Category', 
                        'Sub_Category', 'Avg_Unit_Price', 'Avg_Cost']
product_table['Avg_Unit_Price'] = product_table['Avg_Unit_Price'].round(2)
product_table['Avg_Cost'] = (product_table['Avg_Cost'] / df_cleaned.groupby('Product ID')['Quantity'].mean().values).round(2)

print(f"  ✓ Created Product table with {len(product_table)} unique products")
print(f"  Columns: {product_table.columns.tolist()}")
````

````python
# 3. REGION Table (Reference Data)
print("\n3. Creating REGION table (Reference Data)...")

regions = df_cleaned['Region'].unique()
region_table = pd.DataFrame({
    'Region_ID': range(1, len(regions) + 1),
    'Region_Name': sorted(regions),
    'Region_Code': [r[:3].upper() for r in sorted(regions)]
})

print(f"  ✓ Created Region table with {len(region_table)} regions")
print(f"  Columns: {region_table.columns.tolist()}")
print(f"\n{region_table.to_string(index=False)}")
````

````python
# 4. STATE Table (Reference Data)
print("\n4. Creating STATE table (Reference Data)...")

# Create region lookup
region_lookup = region_table.set_index('Region_Name')['Region_ID'].to_dict()

state_data = df_cleaned.groupby(['State', 'Country/Region', 'Region']).size().reset_index()[['State', 'Country/Region', 'Region']]
state_table = pd.DataFrame({
    'State_ID': range(1, len(state_data) + 1),
    'State_Name': state_data['State'].values,
    'Country': state_data['Country/Region'].values,
    'Region_ID': state_data['Region'].map(region_lookup).values
})

print(f"  ✓ Created State table with {len(state_table)} unique states")
print(f"  Columns: {state_table.columns.tolist()}")
````

````python
# 5. CITY Table (Reference Data)
print("\n5. Creating CITY table (Reference Data)...")

# Create state lookup
state_lookup = state_table.set_index('State_Name')['State_ID'].to_dict()

city_data = df_cleaned.groupby(['City', 'State', 'Postal Code']).size().reset_index()[['City', 'State', 'Postal Code']]
city_table = pd.DataFrame({
    'City_ID': range(1, len(city_data) + 1),
    'City_Name': city_data['City'].values,
    'State_ID': city_data['State'].map(state_lookup).values,
    'Postal_Code': city_data['Postal Code'].values
})

print(f"  ✓ Created City table with {len(city_table)} unique cities")
print(f"  Columns: {city_table.columns.tolist()}")
````

````python
# 6. SHIP_MODE Table (Reference Data)
print("\n6. Creating SHIP_MODE table (Reference Data)...")

ship_modes = df_cleaned.groupby('Ship Mode').agg({
    'Shipping Days': 'mean'
}).reset_index()

ship_mode_table = pd.DataFrame({
    'Ship_Mode_ID': range(1, len(ship_modes) + 1),
    'Ship_Mode_Name': ship_modes['Ship Mode'].values,
    'Avg_Shipping_Days': ship_modes['Shipping Days'].round(1).values
})

print(f"  ✓ Created Ship_Mode table with {len(ship_mode_table)} shipping methods")
print(f"  Columns: {ship_mode_table.columns.tolist()}")
print(f"\n{ship_mode_table.to_string(index=False)}")
````

````python
# 7. REGIONAL_MANAGER Table (Master Data)
print("\n7. Creating REGIONAL_MANAGER table (Master Data)...")

regional_manager_table = pd.DataFrame({
    'Manager_ID': range(1, len(df_people_cleaned) + 1),
    'Manager_Name': df_people_cleaned['Regional Manager'].values,
    'Region_ID': df_people_cleaned['Region'].map(region_lookup).values
})

print(f"  ✓ Created Regional_Manager table with {len(regional_manager_table)} managers")
print(f"  Columns: {regional_manager_table.columns.tolist()}")
print(f"\n{regional_manager_table.to_string(index=False)}")
````

````python
# 8. ORDER_HEADER Table (Transaction Data)
print("\n8. Creating ORDER_HEADER table (Transaction Data)...")

# Create lookups
ship_mode_lookup = ship_mode_table.set_index('Ship_Mode_Name')['Ship_Mode_ID'].to_dict()
city_lookup = city_table.set_index('City_Name')['City_ID'].to_dict()

order_header_data = df_cleaned.groupby('Order ID').agg({
    'Customer ID': 'first',
    'Order Date': 'first',
    'Ship Date': 'first',
    'Ship Mode': 'first',
    'City': 'first',
    'Sales': 'sum'
}).reset_index()

order_header_table = pd.DataFrame({
    'Order_ID': order_header_data['Order ID'].values,
    'Customer_ID': order_header_data['Customer ID'].values,
    'Order_Date': order_header_data['Order Date'].values,
    'Ship_Date': order_header_data['Ship Date'].values,
    'Ship_Mode_ID': order_header_data['Ship Mode'].map(ship_mode_lookup).values,
    'City_ID': order_header_data['City'].map(city_lookup).values,
    'Order_Total': order_header_data['Sales'].round(2).values
})

print(f"  ✓ Created Order_Header table with {len(order_header_table)} unique orders")
print(f"  Columns: {order_header_table.columns.tolist()}")
````

````python
# 9. ORDER_LINE_ITEM Table (Transaction Data)
print("\n9. Creating ORDER_LINE_ITEM table (Transaction Data)...")

order_line_table = pd.DataFrame({
    'Row_ID': df_cleaned['Row ID'].astype(int).values,
    'Order_ID': df_cleaned['Order ID'].values,
    'Product_ID': df_cleaned['Product ID'].values,
    'Quantity': df_cleaned['Quantity'].values,
    'Unit_Price': df_cleaned['Revenue per Unit'].round(2).values,
    'Discount_Percent': df_cleaned['Discount'].round(4).values,
    'Line_Sales': df_cleaned['Sales'].round(2).values,
    'Line_Profit': df_cleaned['Profit'].round(2).values,
    'Line_COGS': df_cleaned['COGS'].round(2).values
})

print(f"  ✓ Created Order_Line_Item table with {len(order_line_table)} line items")
print(f"  Columns: {order_line_table.columns.tolist()}")
````

````python
# 10. RETURN Table (Transaction Data)
print("\n10. Creating RETURN table (Transaction Data)...")

return_table = pd.DataFrame({
    'Return_ID': range(1, len(df_returns_cleaned) + 1),
    'Order_ID': df_returns_cleaned['Order ID'].values,
    'Return_Status': df_returns_cleaned['Returned'].values
})

print(f"  ✓ Created Return table with {len(return_table)} returns")
print(f"  Columns: {return_table.columns.tolist()}")
````

````python
# 11. DATA_DICTIONARY Table (Metadata)
print("\n11. Creating DATA_DICTIONARY table (Metadata)...")

# Create comprehensive data dictionary
metadata_records = []

# Add metadata for each table
tables_metadata = {
    'Customer': customer_table,
    'Product': product_table,
    'Region': region_table,
    'State': state_table,
    'City': city_table,
    'Ship_Mode': ship_mode_table,
    'Regional_Manager': regional_manager_table,
    'Order_Header': order_header_table,
    'Order_Line_Item': order_line_table,
    'Return': return_table
}

for table_name, table_df in tables_metadata.items():
    for col in table_df.columns:
        metadata_records.append({
            'Table_Name': table_name,
            'Column_Name': col,
            'Data_Type': str(table_df[col].dtype),
            'Source_System': 'GBP_DataSource',
            'Last_Updated': datetime.now()
        })

data_dictionary_table = pd.DataFrame(metadata_records)
data_dictionary_table.insert(0, 'Column_ID', range(1, len(data_dictionary_table) + 1))

print(f"  ✓ Created Data_Dictionary table with {len(data_dictionary_table)} metadata entries")
print(f"  Columns: {data_dictionary_table.columns.tolist()}")
````
